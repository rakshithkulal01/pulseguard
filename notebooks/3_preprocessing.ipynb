```python
import numpy as np
import os
```

```python
BASE_PATH = "/content/drive/MyDrive/pluseguard"

AUGMENTED_PATH = os.path.join(BASE_PATH, "Augmented")
SPLIT_PATH = os.path.join(BASE_PATH, "Preprocessed")
NORMALIZED_PATH = os.path.join(BASE_PATH, "Normalized")

os.makedirs(NORMALIZED_PATH, exist_ok=True)

print("Augmented path:", AUGMENTED_PATH)
print("Split path:", SPLIT_PATH)
print("Normalized path:", NORMALIZED_PATH)
```

```python
X_train_augmented = np.load(
    os.path.join(AUGMENTED_PATH, "X_train_augmented.npy")
)

y_train_augmented = np.load(
    os.path.join(AUGMENTED_PATH, "y_train_augmented.npy")
)

print("Augmented Training ECG shape:", X_train_augmented.shape)
print("Augmented Training Labels shape:", y_train_augmented.shape)
```

```python
X_val = np.load(
    os.path.join(SPLIT_PATH, "X_val.npy")
)

y_val = np.load(
    os.path.join(SPLIT_PATH, "y_val.npy")
)

X_test = np.load(
    os.path.join(SPLIT_PATH, "X_test.npy")
)

y_test = np.load(
    os.path.join(SPLIT_PATH, "y_test.npy")
)

print("Validation ECG shape:", X_val.shape)
print("Validation Labels shape:", y_val.shape)

print("Test ECG shape:", X_test.shape)
print("Test Labels shape:", y_test.shape)
```

```python
print("Training:", X_train_augmented.shape)
print("Validation:", X_val.shape)
print("Testing:", X_test.shape)

print()
print("Training labels:", y_train_augmented.shape)
print("Validation labels:", y_val.shape)
print("Testing labels:", y_test.shape)
```

```python
def z_score_normalize(ecg):
    mean = np.mean(ecg)
    std = np.std(ecg)

    if std == 0:
        return np.zeros_like(ecg)

    return (ecg - mean) / std
```

```python
X_train_z = np.array([
    z_score_normalize(ecg)
    for ecg in X_train_augmented
], dtype=np.float32)

print("Normalized training shape:", X_train_z.shape)
```

```python
X_val_z = np.array([
    z_score_normalize(ecg)
    for ecg in X_val
], dtype=np.float32)

print("Normalized validation shape:", X_val_z.shape)
```

```python
X_test_z = np.array([
    z_score_normalize(ecg)
    for ecg in X_test
], dtype=np.float32)

print("Normalized testing shape:", X_test_z.shape)
```

```python
sample_index = 0

print("Before normalization:")
print("Mean:", np.mean(X_train_augmented[sample_index]))
print("Std :", np.std(X_train_augmented[sample_index]))

print("\nAfter normalization:")
print("Mean:", np.mean(X_train_z[sample_index]))
print("Std :", np.std(X_train_z[sample_index]))
```

```python
for i in range(5):
    mean = np.mean(X_train_z[i])
    std = np.std(X_train_z[i])

    print(
        f"ECG {i}: Mean = {mean:.6f}, Std = {std:.6f}"
    )
```

```python
train_means = np.mean(X_train_z, axis=1)
train_stds = np.std(X_train_z, axis=1)

print("Average ECG mean:", np.mean(train_means))
print("Average ECG std :", np.mean(train_stds))

print("Minimum ECG mean:", np.min(train_means))
print("Maximum ECG mean:", np.max(train_means))

print("Minimum ECG std:", np.min(train_stds))
print("Maximum ECG std:", np.max(train_stds))
```

```python
print("Training label distribution:")
unique, counts = np.unique(y_train_augmented, return_counts=True)

for label, count in zip(unique, counts):
    print(f"Label {label}: {count}")

print("\nValidation label distribution:")
unique, counts = np.unique(y_val, return_counts=True)

for label, count in zip(unique, counts):
    print(f"Label {label}: {count}")

print("\nTesting label distribution:")
unique, counts = np.unique(y_test, return_counts=True)

for label, count in zip(unique, counts):
    print(f"Label {label}: {count}")
```

```python
np.save(
    os.path.join(NORMALIZED_PATH, "X_train_z.npy"),
    X_train_z
)

np.save(
    os.path.join(NORMALIZED_PATH, "y_train_z.npy"),
    y_train_augmented
)

print("Normalized training data saved.")
```

```python
np.save(
    os.path.join(NORMALIZED_PATH, "X_val_z.npy"),
    X_val_z
)

np.save(
    os.path.join(NORMALIZED_PATH, "y_val_z.npy"),
    y_val
)

print("Normalized validation data saved.")
```

```python
np.save(
    os.path.join(NORMALIZED_PATH, "X_test_z.npy")
    , X_test_z
)

np.save(
    os.path.join(NORMALIZED_PATH, "y_test_z.npy"),
    y_test
)

print("Normalized testing data saved.")
```
